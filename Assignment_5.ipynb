# WAP in python to implement 8 puzzle problem using BFS

```python
from collections import deque

def find_empty(state):
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                return r, c
    return -1, -1

def get_neighbors(state):
    neighbors = []
    empty_r, empty_c = find_empty(state)

    moves_deltas = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves_deltas:
        new_r, new_c = empty_r + dr, empty_c + dc

        if 0 <= new_r < 3 and 0 <= new_c < 3:
            new_state_list = [list(row) for row in state]

            new_state_list[empty_r][empty_c], new_state_list[new_r][new_c] = \
                new_state_list[new_r][new_c], new_state_list[empty_r][empty_c]

            neighbors.append(tuple(tuple(row) for row in new_state_list))

    return neighbors

def solve_8_puzzle_bfs(initial_state):
    goal_state = ((1, 2, 3), (4, 5, 6), (7, 8, 0))

    queue = deque([(initial_state, [initial_state])])

    visited = {initial_state}

    while queue:
        current_state, path = queue.popleft()

        if current_state == goal_state:
            return path

        for neighbor_state in get_neighbors(current_state):
            if neighbor_state not in visited:
                visited.add(neighbor_state)
                new_path = path + [neighbor_state]
                queue.append((neighbor_state, new_path))

    return None

def print_puzzle(state):
    for row in state:
        print(" ".join(map(str, row)))
    print()

print("--------------------------------------")
print("Solving 8-puzzle with BFS (no heuristics)")
print("--------------------------------------")

initial_puzzle = (
    (1, 0, 2),
    (3, 4, 5),
    (6, 7, 8)
)

print("Initial State:")
print_puzzle(initial_puzzle)

solution_path = solve_8_puzzle_bfs(initial_puzzle)

if solution_path:
    print(f"Solution found in {len(solution_path) - 1} steps:")
    for i, state in enumerate(solution_path):
        print(f"Step {i}:")
        print_puzzle(state)
else:
    print("No solution found for the given initial state.")

print("\n--------------------------------------")
print("Another example:")
print("--------------------------------------")
initial_puzzle_2 = (
    (8, 6, 7),
    (2, 5, 4),
    (3, 0, 1)
)

print("Initial State 2:")
print_puzzle(initial_puzzle_2)

solution_path_2 = solve_8_puzzle_bfs(initial_puzzle_2)

if solution_path_2:
    print(f"Solution found for State 2 in {len(solution_path_2) - 1} steps:")
else:
    print("No solution found for initial state 2.")
```
