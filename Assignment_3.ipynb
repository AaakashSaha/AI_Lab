```python
# import libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Load the dataset from a valid public URL
url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv'
iris = pd.read_csv(url)

# Display dataset info
print("Shape of the dataset:", iris.shape)
print("\nFirst 5 Rows:")
print(iris.head())
```

```python
#select Features (X) and Target (Y)
x = iris.iloc[:, 0:4] # Features: sepal_length, sepal_width, petal_length, petal_width
y = iris["species"] # Target Column (lowercase 'species')

#Split the dataset into Training and Testing sets
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.8, random_state=42
)

#Create the KNN model
knn = KNeighborsClassifier(n_neighbors=3)
#train the model
knn.fit(x_train, y_train)
#predict on test data
y_pred = knn.predict(x_test)

#Display
print("\nPredicted Species: ")
print(y_pred)

print("\nActual Species:")
print(y_test.values)

#Calculate Accuracy
accuracy = accuracy_score(y_test, y_pred)
print("\nAccuracy:", accuracy * 100, "%")

#Confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

#classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```
