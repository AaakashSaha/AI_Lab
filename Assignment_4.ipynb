# BFS

```python
from collections import deque

graph={
    'A':['B','C'],
    'B':['D','E'],
    'C':['F'],
    'D':[],
    'E':['G'],
    'F':[],
    'G':[]
}
def bfs(graph,start):
  visited=set()
  queue=deque([start])
  traversal=[]
  visited.add(start)
  while queue:
    node=queue.popleft()
    traversal.append(node)
    for neighbour in graph[node]:
      if neighbour not in visited:
        visited.add(neighbour)
        queue.append(neighbour)
  return traversal
#run  BFS
result=bfs(graph,'A')
print("BFS Traversal : ")
print("->".join(result))

```

```python
import networkx as nx
import matplotlib.pyplot as plt
G=nx.DiGraph()
for node in graph:
  for neighbour in graph[node]:
    G.add_edge(node,neighbour)
pos={
    'A':(0,0),
    'B':(1,0),
    'C':(2,0),
    'D':(0,1),
    'E':(1,1),
    'F':(2,1),
    'G':(1,2)
}
plt.figure(figsize=(8,6))
nx.draw(G,pos,with_labels=True,node_size=2500,node_color='skyblue',font_size=14,arrowsize=20,font_weight='bold',edge_color='gray',arrows=True,width=2)
plt.title("BFS Traversal")
plt.show()
```

# DFS


```python
graph = {
    'A':['B','C'],
    'B':['D','E'],
    'C':['F'],
    'D':[],
    'E':['G'],
    'F':[],
    'G':[]
}

def dfs(graph, start, visited=None, traversal=None):
    if visited is None:
        visited = set()
    if traversal is None:
        traversal = []

    visited.add(start)
    stack = [start]

    while stack:
        node = stack.pop()
        traversal.append(node)

        for neighbour in reversed(graph[node]):
            if neighbour not in visited:
                visited.add(neighbour)
                stack.append(neighbour)

    return traversal

# Run DFS
result = dfs(graph, 'A')
print("DFS Traversal : ")
print("->".join(result))
```
